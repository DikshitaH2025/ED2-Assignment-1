## Q.2) Also considering acceleration along i)0.3k and ii)0.9x and plot the field lines as  fucntion of time at a fixed distance far away from the source.

In this q, we will be considering only the radiation field as we are at a far away position. So we also have to consider a retarded time.

#### For accleration in the x direction (0.9x)

```python
#Defining constants
x=30e8
y=30e8
c=3e8
r_obs=np.array([x,y,0]) #Vector defining position of E/B observation
r_c=np.array([0,0,0]) #Vector defining the initial position of the charge

t_i=np.sqrt(np.dot(r_obs,r_obs))/c
t=np.linspace(t_i,t_i+100,1000) #Time axis

#beta fucntions
v_x=0.1
b_o=np.array([v_x,0,0]) #velocity

aaa=0.3
a_x=0
a_y=0
a_z=0.3
B=np.array([a_x,a_y,a_z]) #acceleration


#Kappa Function
def K(n,b):
    kappa=1-np.dot(n,b)
    return kappa

    
#Defining the retarded time array
t_r=np.zeros(len(t))

#Loop to find the retarded time array
ii=0
while ii<len(t):
    r_c=(b_o*t[ii])+(0.5*B*t[ii]*t[ii]) #New instantaneous position
    R=np.sqrt(np.dot(r_obs-r_c,r_obs-r_c)) #Distance between the charge and obs pt
    t_r[ii]=t[ii]-R/c
    ii=ii+1

#Array for the position of charge in retarded time (This will evolve over time)
R_cx=b_o[0]*t_r+0.5*B[0]*t_r**2 #New x position
R_cy=b_o[1]*t_r+0.5*B[1]*t_r**2 #New y position
R_cz=b_o[2]*t_r+0.5*B[2]*t_r**2 #New z position


#Defining the displacement vectors
Rxx=r_obs[0]-R_cx
Ryy=r_obs[1]-R_cy
Rzz=r_obs[2]-R_cz


#Distance of the points
D=[]
D=(Rxx*Rxx)+(Ryy*Ryy)+(Rzz*Rzz)
D[D == 0] = np.nan

#Initialting the unit vector arrays
nx=np.zeros(len(D))
ny=np.zeros(len(D))
nz=np.zeros(len(D))


#Array for the unit direction 
ii=0
while ii<len(D):
    nx[ii]=Rxx[ii]/m.sqrt(D[ii])
    ny[ii]=Ryy[ii]/m.sqrt(D[ii])
    nz[ii]=Rzz[ii]/m.sqrt(D[ii])
    ii=ii+1


#Defining the electric field array
E=np.zeros((len(t),3))

ii=0
while ii<len(D):
    n=np.array([nx[ii],ny[ii],nz[ii]])
    #print(f"For {ii} we have n={n}")
    v1=np.cross((n-b_o),B)
    v2=np.cross(n,v1)
    #print(f"For {ii} we have v1={v1}, v2={v2} and v3={v3}")
    E[ii,:]=q*v2/((K(n,b_o)**3)*D[ii]) #Electric field value at each point
    ii=ii+1

Ex=E[:,0]
Ey=E[:,1]
Ez=E[:,2]

plt.plot(Ex,t)
plt.xlabel("Time axis")
plt.ylabel("Electric Field along x-direction")
plt.title(f"Electric Filed vs Time (x-axis) for beta={v_x} and a={aaa}z")
plt.grid()
plt.savefig('q2_b=0.1c_a=0.3z_Ex.jpg', dpi=100)
plt.show()

plt.plot(Ey,t)
plt.xlabel("Time axis")
plt.ylabel("Electric Field along y-direction")
plt.title(f"Electric Filed vs Time (y-axis) for beta={v_x} and a={aaa}z")
plt.grid()
plt.savefig('q2_b=0.1c_a=0.3z_Ey.jpg', dpi=100)
plt.show()

plt.plot(Ez,t)
plt.xlabel("Time axis")
plt.ylabel("Electric Field along z-direction")
plt.title(f"Electric Filed vs Time (z-axis) for beta={v_x} and a={aaa}z")
plt.grid()
plt.savefig('q2_b=0.1c_a=0.3z_Ez.jpg', dpi=100)
plt.show()


#Defining the Magnetic field array
BB=np.zeros((len(t),3))

ii=0
while ii<len(t_r):
    E_B=np.array([Ex[ii],Ey[ii],Ez[ii]])
    n=np.array([nx[ii],ny[ii],nz[ii]])
    BB[ii,:]=np.cross(n,E_B)
    ii=ii+1

Bx=BB[:,0]
By=BB[:,1]
Bz=BB[:,2]


plt.plot(Bx,t)
plt.xlabel("Time axis")
plt.ylabel("Magnetic Field along x-direction")
plt.title(f"Magnetic Filed vs Time (x-axis) for beta={v_x} and a={aaa}z")
plt.grid()
plt.savefig('q2_b=0.1c_a=0.3z_Bx.jpg', dpi=100)
plt.show()

plt.plot(By,t)
plt.xlabel("Time axis")
plt.ylabel("Magnetic Field along y-direction")
plt.title(f"Magnetic Filed vs Time (y-axis) for beta={v_x} and a={aaa}z")
plt.grid()
plt.savefig('q2_b=0.1c_a=0.3z_By.jpg', dpi=100)
plt.show()

plt.plot(Bz,t)
plt.xlabel("Time axis")
plt.ylabel("Magnetic Field along z-direction")
plt.title(f"Magnetic Filed vs Time (z-axis) for beta={v_x} and a={aaa}z")
plt.grid()
plt.savefig('q2_b=0.1c_a=0.3z_Bz.jpg', dpi=100)
plt.show()
```

## Q3) Plotting the angular distribution of the radiated power by the particle

This q requires us to make a plot the distribution of power vs theta , now we will have to use the spherical coordinate system and we will work with a fixed R and varying theta and phi angle that will be incorporated in the n.


#### Angular Distribution of Radiated Power for the accelration along x direction (0.9x)

```python
#The angle arrays
theta = np.linspace(0, np.pi, 100) 
phi = np.linspace(0, 2*np.pi, 100)

#Creating a meshgrid
T,P=np.meshgrid(theta, phi)

#defining the unit vector n=(sinTcosP,sinTsinP,cosT)
nx = np.sin(T) * np.cos(P)
ny = np.sin(T) * np.sin(P)
nz = np.cos(T)

#Defining the position of particle
R=30e8

c=3e8 #Speed of light

#beta fucntions
v_x=0.999
v_y=0
v_z=0
b_o=np.array([v_x,v_y,v_z]) #velocity

a_x=0.999
a_y=0
a_z=0
B=np.array([a_x,a_y,a_z]) #acceleration

#Kappa Function
def K(n,b):
    kappa=1-np.dot(n,b)
    return kappa

#DEFINING THE ELECTRIC FIELD
Power = np.zeros((len(theta),len(theta))) #The radiated power    

i=0
while i<len(theta):
    j=0
    while j<len(phi):
        n=np.array([nx[i][j],ny[i][j],nz[i][j]])
        v1=np.cross((n-b_o),B)
        v2=np.cross(n,v1)
        E=q*v2/((K(n,b_o)**3)*R)
        Power[i][j]=np.dot(E,E)*c/R**2
        j=j+1
    i=i+1

plt.contourf(T, P, Power, levels=30, cmap='RdBu')
plt.colorbar(label='Angular distribution of power')
plt.xlabel('Theta (radians)')
plt.ylabel('Phi (radians)')
plt.title(f'Angular Distributon of Power beta={v_x}c and a=0.9x')
plt.savefig('q3_b=0.999c_a=0.9x_P_Dist.jpg', dpi=100)
plt.show()
```

## Q6) Power distribution for a given velocity distribution

```python
#Package used
from scipy.integrate import quad
import math as m

#Defining the constants
mm=9.1e-31 #Mass of the electron
q=1.6e-19 #charge of the electron
c=3e8 #velocity of light
kb=1.38e-23 #Boltzmann Constant
h=6.626e-34 #Planck's constant

#System related parameters
Z=70
ne=1 #Density of electrons
ni=1 #Density of ions

#Constant for the power of each individual particle
CC=(16*m.pi*(q**6)*Z*Z*ne*ni)/(3*m.sqrt(3)*mm*c**3)


#Defining the velocity arrays
V=np.linspace(1e-2,3000,100)


#Fucntion for the min velocity
def v_min(nu):
    a=m.sqrt((2*h*nu)/mm)
    return a


#Defining the velocity distribution
def f(v):
    ct=mm*v*v*0.5/(kb*T)
    FF=v**4*np.exp(-ct)*(1+(0.3*0.5*ct))
    return FF


#Defining a freq array
fq=np.linspace(1e10,1e14,100000)

#Defining the Gaunt factor
def g_f(nu, T):
    xx = ((2*kb*T)**1.5) / (np.pi *q**2 * h * nu * np.sqrt(mm))
    gff = (np.sqrt(3)/np.pi) * np.log(xx + np.e)  # prevent negative inside log
    return np.maximum(gff, 1)
    
#Defining the Numerator integral
def Nf(v):
    I=CC*f(v)/v
    return I
```

```python
#Defining Power array
P=np.zeros(len(fq))

T=300
#Charcateristic Freq
nu_peak = kb*T/h
print(f"Characteristic frequency:{nu_peak}")

#Finding the power as a function of freq
ii=0
while ii<len(fq):
    vmin=v_min(fq[ii])
    vmax=vmin*6 #Setting a max for the normnalization factor as our function converges
    g=g_f(fq[ii],T)
    Denom, err_D = quad(f, 0,vmax,limit=7)
    FF, err_f = quad(Nf, vmin,vmax,limit=10)
    P[ii]=FF*g/Denom
    ii=ii+1
plt.semilogx(fq,P) #Plotting the Power vs Freq in semi-log


#Plotting Power vs Frequency
plt.xlabel("Frequency")
plt.ylabel("Power per unit Frequency")
plt.title(f"Variation of Power wrt Frequency at constant T=300K")
plt.grid()
plt.savefig("P vs Freq (T=300K).jpg",dpi=100)
plt.show()
```
