### Q.1.b) Plotting the Extra Terms in Maxwell Eqns due to the Galiliean Transformation

In this we have considered the velocity to be along only the +ve x-axis direction and since the terms are only dot products of the field compoennt and the elcovity we are taking only the x components of teg E and B

```python

# Making the grid
x = np.linspace(-5, 5, 100)

#Velocity (motion only along +ve x-axis)
v=np.array([0.01,0.5,0.9999])

#Charge density (taking it as constant)
rho=20 #Increased to show its effect wrt the magnetic field extra term


#Defining our Electric and Magnetic Field
Ex = x**3

Bx = x**3

# Finding the gradients of the E and B 
dEx_dx= np.gradient(Ex, x)

dBx_dx= np.gradient(Bx, x)


# Computing the extra term - 
# Maxwell's 3rd Eqn :
for i in v:
    ext_T3=-rho*i + i*dEx_dx
    plt.plot(x,ext_T3)

plt.xlabel("x-axis")
plt.ylabel("Extra Term (-rho*v + v.Del E)")
plt.title("Variation of the Extra Term in 3rd Maxwell Eqn with varying v")
plt.legend(['v/c=0.01','v/c=0.5','v/c=0.9999'])
plt.grid()
plt.savefig('q1b_fig1.jpg', dpi=100)

plt.show()

# Maxwell's 4th Eqn :
for i in v:
    ext_T4 = i*dBx_dx
    plt.plot(x,ext_T4)

plt.xlabel("x-axis")
plt.ylabel("Extra Term (v.Del B)")
plt.title("Variation of the Extra Term in 4th Maxwell Eqn with varying v")
plt.legend(['v/c=0.01','v/c=0.5','v/c=0.9999'])
plt.grid()
plt.savefig('q1b_fig2.jpg', dpi=100)

plt.show()
```

### Q.3. Plot for elementary particle with lifetime t 

```python
#User-defined func for gamma value 
def gamma_v(v):
    gamma=1/m.sqrt(1-v*v)
    return gamma

t=np.arange(1,100)
t=t*10**(-6)

t_p=gamma_v(0.01)*t
plt.plot(t,t_p)

t_p=gamma_v(0.5)*t
plt.plot(t,t_p)

t_p=gamma_v(0.9999)*t
plt.plot(t,t_p)

plt.xlabel("Particle Rest lifetime (µs)")
plt.ylabel("Lab lifetime (µs)")
plt.title("Time dilation of lifetime")
plt.legend(['v/c=0.01','v/c=0.5','v/c=0.9999'])
plt.savefig('q3_fig1.jpg', dpi=100)
plt.show()
```

```python
t=1e-6 # rest lifetime

v = np.linspace(0, 0.9999, 100)  # speeds as fraction of c
v=np.append(v,[0.01,0.5])
v=np.sort(v)

gamma=1/np.sqrt(1-v**2)
t_l = gamma * t

plt.plot(v, t_l*1e6)
plt.xlabel("v/c")
plt.ylabel("Lab lifetime τ′ (µs)")

plt.scatter(0.01,1.0000500037503126, color='red', s=80, zorder=5, label='v/c=0.01')
plt.scatter(0.5,1.1547005383792516, color='green', s=80, zorder=5, label='v/c=0.5')
plt.scatter(0.9999,70.71244595191451, color='orange', s=80, zorder=5, label='v/c=0.9999')

plt.legend()
plt.title("Time dilation of lifetime")

plt.savefig('q3_fig2.jpg', dpi=100)
plt.show()
```

### Q. 2 a) Space-Time Diagram for various velocities

```python
#That mesh gave a good plot but we need to make sure it doesn't leak into the 
#older coordinate

#For the tick marks and the smaller lines of the ST diagram
x_v=np.linspace(-20,20,60)
t_v=np.linspace(-20,20,60)
t, x = np.meshgrid(t_v, x_v)   

def gamma_v(v):
    g=1/m.sqrt(1-v*v)
    return g

#User-defined function to get the new coordinates 
def Lorentz_T(x,t,v):
    O=m.atanh(v)
    t_p=m.cosh(O)*t-m.sinh(O)*x
    x_p=-m.sinh(O)*t+m.cosh(O)*x
    return t_p,x_p

#The main axes
plt.axvline(0,color='black')
plt.axhline(0,color='black')

#Plotting for the light axis
plt.plot(x_v,x_v,color='black',linestyle='--',linewidth=0.5,label='light axis')
plt.plot(x_v,-x_v,color='black',linewidth=0.5,linestyle='--')


#Plotting the Space-Time Diagram
v=0.01

plt.plot(x_v,v*t_v,'b',linewidth=1,label='New Axes (v=0.01)')
plt.plot(v*x_v,t_v,'b',linewidth=1)

t_1,x_1=Lorentz_T(x,t,v)

plt.plot(x_1,t,'-b',linewidth=0.2)
plt.plot(x,t_1,'-b',linewidth=0.2)


plt.xlabel("x-axis")
plt.ylabel("t-axis")
plt.legend()

plt.savefig('q2_a_0.01.jpg', dpi=100)

plt.title("Space-Time Diagram")
plt.show()
```
